```python
import pandas as pd

from datastore360.clean import (
    normalize_text_columns, parse_dates, fix_typos, remove_duplicates,
    fix_known_row_id_conflicts, handle_missing_ship_date_mode,
    handle_missing_sales_quantity, handle_missing_postal_code,
    handle_missing_customer_name, validate_discount, validate_quantity,
    hash_customer_name, add_delivery_time, add_profit_margin
)

df = pd.read_csv('../data/raw/store-data-360.csv')
```

```python
df = normalize_text_columns(df)
```

```python
df = parse_dates(df)
```

```python
df = fix_typos(df)
```

```python
df = fix_known_row_id_conflicts(df)
```

```python
df = handle_missing_ship_date_mode(df)
```

```python
df = handle_missing_sales_quantity(df)
```

```python
df = handle_missing_postal_code(df)
```

```python
df = handle_missing_customer_name(df)
```

```python
df = validate_discount(df)
```

```python
df = validate_quantity(df)
```

```python
df = remove_duplicates(df)
```

```python
df = add_delivery_time(df)
```

```python
df = add_profit_margin(df)
```

```python
df = hash_customer_name(df)
```

```python
df.to_csv("../data/processed/superstore_clean.csv", index=False)
print("Final shape:", df.shape)
```
